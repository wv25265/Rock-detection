```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.train(
      data=f"{dataset.location}/data.yaml",
      epochs=100,
      imgsz=640,
      batch=16,
      patience=20,
      name="rock_v8n"
  )
```

```python
results = model.predict(f"{dataset.location}/test/images", save=True,
  conf=0.25)
print("results saved at:", results[0].save_dir)
```
